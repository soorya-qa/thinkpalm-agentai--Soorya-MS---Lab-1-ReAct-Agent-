```python
# -------- Tool --------
def calculator(expression):
    return eval(expression)

# -------- ReAct Agent --------
def react_agent(user_query):
    print(f"User Query: {user_query}\n")

    # Reasoning
    print("Thought: The user is asking how long their phone battery will last.")
    print("Thought: The phone has 80% battery remaining.")
    print("Thought: The battery drains at 10% per hour.")
    print("Thought: To find time, I should divide remaining battery by drain rate.\n")

    # Action (tool usage)
    expression = "80 / 10"
    print(f"Action: calculator('{expression}')")
    result = calculator(expression)

    # Observation
    print(f"Observation: {result} hours\n")

    # Final Answer
    print(f"Final Answer: The phone battery will last approximately {int(result)} hours.")

# -------- Run the agent --------
query = "If my phone battery is 80% and it drains 10% per hour, how long will it last?"
react_agent(query)
```
